```python
import torch
import torch.nn as nn
import torch.nn.functional as f
```

```python
class nnetwork(nn.Module):
    def __init__(self,i_size,o_size):
        super(nnetwork,self).__init__()
        self.l1=nn.Linear(i_size,256)
        self.l2=nn.Linear(256,64)
        self.l3=nn.Linear(64,o_size)
        self.bn1=nn.BatchNorm1d(256)
        self.bn2=nn.BatchNorm1d(64)
        self.drop=nn.Dropout(0.2)
    
    def forward(self,x):
        x=x.view(x.size(0),-1)
        x=f.relu(self.bn1(self.l1(x)))
        x=self.drop(x)
        x=f.relu(self.bn2(self.l2(x)))
        x=self.drop(x)
        x=self.l3(x)
        return x
```

```python
batch_size=64
input_size=784
output_size=10
num_epochs=15
learning_rate=0.001
model=nnetwork(input_size,output_size)
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=learning_rate)
```

```python
import keras
from keras.datasets import mnist
```

```python
(x_train,y_train),(x_test,y_test)=mnist.load_data()
```

```python
from torch.utils.data import DataLoader,Dataset
class MNISTDataset(Dataset):
    def __init__(self,images,labels):
        self.images=images
        self.labels=labels
    def __len__(self):
        return len(self.images)
    def __getitem__(self,idx):
        image=self.images[idx]
        label=self.labels[idx]

        image = torch.from_numpy(image).float() / 255.0
        label = torch.tensor(label).long()

        return image,label
```

```python
train_data = DataLoader(MNISTDataset(x_train,y_train),batch_size=64,shuffle=True)  
test_data = DataLoader(MNISTDataset(x_test,y_test),batch_size=64,shuffle=True)
```

```python
for epoch in range(num_epochs):
    model.train()
    i=0
    for images,labels in train_data:
        outputs=model(images)
        loss=criterion(outputs,labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        print(i)
        i+=1
    print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')
```

```python
correct=0
total=0
model.eval()
for images,labels in test_data:
    outputs=model(images)
    loss=criterion(outputs,labels)
    correct+=(outputs.argmax(1)==labels).sum().item()
    total+=labels.size(0)
print(f'Test Accuracy: {100*correct/total:.2f}%')
```
